```python
# ============================================================
# CELL 6
# PATIENT-WISE TRAIN / VALIDATION / TEST SPLIT
# ============================================================

from sklearn.model_selection import GroupShuffleSplit


def patient_wise_split(
    dataframe,
    patient_column="Patient ID",
    test_fraction=0.15,
    validation_fraction=0.15,
    random_state=42
):
    """
    Patient-wise splitting ensures that X-rays from the same
    patient cannot appear in more than one subset.
    """

    # --------------------------------------------------------
    # First split:
    # 85% Train+Validation / 15% Test
    # --------------------------------------------------------

    first_split = GroupShuffleSplit(
        n_splits=1,
        test_size=test_fraction,
        random_state=random_state
    )

    train_val_idx, test_idx = next(
        first_split.split(
            dataframe,
            groups=dataframe[patient_column]
        )
    )

    train_val_df = (
        dataframe
        .iloc[train_val_idx]
        .copy()
        .reset_index(drop=True)
    )

    test_pool_df = (
        dataframe
        .iloc[test_idx]
        .copy()
        .reset_index(drop=True)
    )

    # --------------------------------------------------------
    # Second split:
    # Separate validation from training
    # --------------------------------------------------------

    adjusted_val_fraction = (
        validation_fraction /
        (1 - test_fraction)
    )

    second_split = GroupShuffleSplit(
        n_splits=1,
        test_size=adjusted_val_fraction,
        random_state=random_state
    )

    train_idx, val_idx = next(
        second_split.split(
            train_val_df,
            groups=train_val_df[patient_column]
        )
    )

    train_pool_df = (
        train_val_df
        .iloc[train_idx]
        .copy()
        .reset_index(drop=True)
    )

    validation_pool_df = (
        train_val_df
        .iloc[val_idx]
        .copy()
        .reset_index(drop=True)
    )

    return (
        train_pool_df,
        validation_pool_df,
        test_pool_df
    )


# ============================================================
# CREATE PATIENT-WISE POOLS
# ============================================================

(
    train_pool_df,
    validation_pool_df,
    test_pool_df
) = patient_wise_split(
    dataframe=df,
    patient_column="Patient ID",
    test_fraction=0.15,
    validation_fraction=0.15,
    random_state=SEED
)


# ============================================================
# DISPLAY POOL SIZES
# ============================================================

print("=" * 60)
print("PATIENT-WISE SOURCE POOLS")
print("=" * 60)

print(
    "Training pool:",
    len(train_pool_df)
)

print(
    "Validation pool:",
    len(validation_pool_df)
)

print(
    "Testing pool:",
    len(test_pool_df)
)

print(
    "Total:",
    len(train_pool_df)
    + len(validation_pool_df)
    + len(test_pool_df)
)


# ============================================================
# VERIFY PATIENT SEPARATION
# ============================================================

train_patients = set(
    train_pool_df["Patient ID"]
)

validation_patients = set(
    validation_pool_df["Patient ID"]
)

test_patients = set(
    test_pool_df["Patient ID"]
)


train_val_overlap = (
    train_patients &
    validation_patients
)

train_test_overlap = (
    train_patients &
    test_patients
)

val_test_overlap = (
    validation_patients &
    test_patients
)


print("\nPatient leakage check:")

print(
    "Train ↔ Validation:",
    len(train_val_overlap)
)

print(
    "Train ↔ Test:",
    len(train_test_overlap)
)

print(
    "Validation ↔ Test:",
    len(val_test_overlap)
)


assert len(train_val_overlap) == 0
assert len(train_test_overlap) == 0
assert len(val_test_overlap) == 0

print(
    "\n✓ NO PATIENT LEAKAGE DETECTED"
)


# ============================================================
# CHECK AVAILABLE POSITIVES IN EACH POOL
# ============================================================

pool_distribution = pd.DataFrame({
    "Disease": CLASS_NAMES,

    "Training Pool": [
        int(train_pool_df[disease].sum())
        for disease in CLASS_NAMES
    ],

    "Validation Pool": [
        int(validation_pool_df[disease].sum())
        for disease in CLASS_NAMES
    ],

    "Testing Pool": [
        int(test_pool_df[disease].sum())
        for disease in CLASS_NAMES
    ]
})

display(pool_distribution)
```

```python
# ============================================================
# CELL 7
# BALANCED REPRESENTATIVE SAMPLING
# FINAL 10,000 X-RAYS
# ============================================================

import numpy as np
import pandas as pd


def balanced_multilabel_sample(
    dataframe,
    sample_size,
    class_names,
    random_state=42
):
    """
    Selects exactly sample_size UNIQUE X-rays while attempting
    to balance positive disease representation.

    The algorithm gives priority to:
    1. underrepresented diseases,
    2. multi-label images,
    3. patient diversity,
    4. rare disease representation.

     It also penalizes already overrepresented diseases.
    """

    rng = np.random.default_rng(random_state)

    df_local = (
        dataframe
        .copy()
        .reset_index(drop=True)
    )

    if sample_size > len(df_local):
        raise ValueError(
            f"Requested {sample_size} images, "
            f"but only {len(df_local)} are available."
        )

    labels = (
        df_local[class_names]
        .values
        .astype(np.int32)
    )

    num_classes = len(class_names)

    # --------------------------------------------------------
    # Realistic target based on multilabel density
    # --------------------------------------------------------

    total_positive_occurrences = labels.sum()

    average_labels_per_image = (
        total_positive_occurrences /
        max(len(df_local), 1)
    )

    target_positive_per_class = (
        sample_size *
        average_labels_per_image /
        num_classes
    )

    print(
        "\nRequested images:",
        sample_size
    )

    print(
        "Average positive labels/image:",
        round(average_labels_per_image, 3)
    )

    print(
        "Approximate target positives/class:",
        round(target_positive_per_class, 2)
    )

    # --------------------------------------------------------
    # Rarity weights
    # --------------------------------------------------------

    available_counts = (
        labels.sum(axis=0)
    )

    rarity_weights = (
        1.0 /
        np.maximum(
            available_counts,
            1
        )
    )

    rarity_weights = (
        rarity_weights /
        rarity_weights.max()
    )

    # --------------------------------------------------------
    # Information score
    # --------------------------------------------------------

    label_count = labels.sum(axis=1)

    rare_score = (
        labels *
        rarity_weights.reshape(1, -1)
    ).sum(axis=1)

    multilabel_bonus = (
        np.minimum(label_count, 3)
        * 0.05
    )

    importance_score = (
        rare_score
        + multilabel_bonus
    )

    # --------------------------------------------------------
    # Patient information
    # --------------------------------------------------------

    patient_ids = (
        df_local["Patient ID"]
        .astype(str)
        .values
    )

    selected = []
    selected_set = set()

    current_counts = np.zeros(
        num_classes,
        dtype=np.int32
    )

    patient_counter = {}

    remaining = np.arange(
        len(df_local),
        dtype=np.int32
    )

    # --------------------------------------------------------
    # Greedy selection
    # --------------------------------------------------------

    while len(selected) < sample_size:

        candidate_labels = labels[
            remaining
        ]

        # How far each disease is below target
        deficit = np.maximum(
            target_positive_per_class
            - current_counts,
            0
        )

        deficit_weight = (
            deficit /
            max(target_positive_per_class, 1)
        )

        balance_gain = (
            candidate_labels *
            deficit_weight.reshape(1, -1)
        ).sum(axis=1)

        # Penalize diseases that already exceed target
        excess = np.maximum(
            current_counts
            - target_positive_per_class,
            0
        )

        excess_weight = (
            excess /
            max(target_positive_per_class, 1)
        )

        excess_penalty = (
            candidate_labels *
            excess_weight.reshape(1, -1)
        ).sum(axis=1)

        # Strong penalty when a disease becomes too dominant
        upper_limit = (
            target_positive_per_class
            * 1.08
        )

        hard_excess = (
            current_counts
            >= upper_limit
        ).astype(np.float32)

        hard_penalty = (
            candidate_labels *
            hard_excess.reshape(1, -1)
        ).sum(axis=1)

        # Patient diversity
        patient_penalty = np.array(
            [
                patient_counter.get(
                    patient_ids[index],
                    0
                )
                for index in remaining
            ],
            dtype=np.float32
        )

        # Final score
        scores = (
            10.0 * balance_gain
            + 1.0 * importance_score[remaining]
            - 5.0 * excess_penalty
            - 8.0 * hard_penalty
            - 0.03 * patient_penalty
        )

        # Small random tiebreaker
        scores += (
            rng.random(len(scores))
            * 1e-6
        )

        best_position = int(
            np.argmax(scores)
        )

        best_index = int(
            remaining[best_position]
        )

        selected.append(
            best_index
        )

        selected_set.add(
            best_index
        )

        current_counts += (
            labels[best_index]
        )

        patient = (
            patient_ids[best_index]
        )

        patient_counter[patient] = (
            patient_counter.get(
                patient,
                0
            )
            + 1
        )

        remaining = np.delete(
            remaining,
            best_position
        )

    # --------------------------------------------------------
    # Final selected dataframe
    # --------------------------------------------------------

    selected_df = (
        df_local
        .iloc[selected]
        .copy()
        .sample(
            frac=1,
            random_state=random_state
        )
        .reset_index(drop=True)
    )

    return selected_df


# ============================================================
# CREATE FINAL 10,000-IMAGE DATASET
# ============================================================

train_df = balanced_multilabel_sample(
    dataframe=train_pool_df,
    sample_size=7000,
    class_names=CLASS_NAMES,
    random_state=42
)

validation_df = balanced_multilabel_sample(
    dataframe=validation_pool_df,
    sample_size=1500,
    class_names=CLASS_NAMES,
    random_state=43
)

test_df = balanced_multilabel_sample(
    dataframe=test_pool_df,
    sample_size=1500,
    class_names=CLASS_NAMES,
    random_state=44
)


# ============================================================
# FINAL SIZE CHECK
# ============================================================

print("\n" + "=" * 60)
print("FINAL DATASET SIZE")
print("=" * 60)

print(
    "Training:",
    len(train_df)
)

print(
    "Validation:",
    len(validation_df)
)

print(
    "Testing:",
    len(test_df)
)

total_selected = (
    len(train_df)
    + len(validation_df)
    + len(test_df)
)

print(
    "Total unique X-rays:",
    total_selected
)

assert total_selected == 10000

print(
    "\n✓ EXACTLY 10,000 UNIQUE X-RAYS SELECTED"
)
```

```python
# ============================================================
# CELL 11
# IMAGE PREPROCESSING + TENSORFLOW DATASETS
# ============================================================

import cv2
import tensorflow as tf
import numpy as np

from tensorflow.keras.applications.densenet import preprocess_input


# ------------------------------------------------------------
# CLAHE PREPROCESSING
# ------------------------------------------------------------

def clahe_preprocess_numpy(image_path_bytes):
    """
    Reads chest X-ray,
    converts to grayscale,
    applies CLAHE,
    resizes to 224x224,
    converts to RGB,
    applies DenseNet121 preprocess_input.
    """

    image_path = image_path_bytes.decode("utf-8")

    image = cv2.imread(
        image_path,
        cv2.IMREAD_GRAYSCALE
    )

    if image is None:
        raise ValueError(
            f"Unable to read image: {image_path}"
        )

    # CLAHE
    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    image = clahe.apply(image)

    # Resize
    image = cv2.resize(
        image,
        (IMAGE_WIDTH, IMAGE_HEIGHT)
    )

    # Convert grayscale → RGB
    image = cv2.cvtColor(
        image,
        cv2.COLOR_GRAY2RGB
    )

    image = image.astype(
        np.float32
    )

    # DenseNet121 normalization
    image = preprocess_input(image)

    return image


# ------------------------------------------------------------
# TensorFlow wrapper
# ------------------------------------------------------------

def tensorflow_preprocess(
    image_path,
    labels
):

    image = tf.numpy_function(
        func=clahe_preprocess_numpy,
        inp=[image_path],
        Tout=tf.float32
    )

    image.set_shape(
        (
            IMAGE_HEIGHT,
            IMAGE_WIDTH,
            3
        )
    )

    labels = tf.cast(
        labels,
        tf.float32
    )

    labels.set_shape(
        (NUM_CLASSES,)
    )

    return image, labels


# ------------------------------------------------------------
# CREATE TF.DATA DATASET
# ------------------------------------------------------------

def create_tf_dataset(
    dataframe,
    training=False
):

    image_paths = (
        dataframe["image_path"]
        .astype(str)
        .values
    )

    labels = (
        dataframe[CLASS_NAMES]
        .values
        .astype(np.float32)
    )

    dataset = tf.data.Dataset.from_tensor_slices(
        (
            image_paths,
            labels
        )
    )

    if training:
        dataset = dataset.shuffle(
            buffer_size=min(
                len(dataframe),
                7000
            ),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        tensorflow_preprocess,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(
        BATCH_SIZE
    )

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


# ------------------------------------------------------------
# CREATE TRAIN / VALIDATION / TEST DATASETS
# ------------------------------------------------------------

train_dataset = create_tf_dataset(
    train_df,
    training=True
)

validation_dataset = create_tf_dataset(
    validation_df,
    training=False
)

test_dataset = create_tf_dataset(
    test_df,
    training=False
)


print("TensorFlow datasets created successfully.")
```

```python
# ============================================================
# CELL 13
# BUILD DENSENET121 - 13 DISEASE CLASSES
# ============================================================

import tensorflow as tf

from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.layers import (
    GlobalAveragePooling2D,
    Dense,
    Dropout
)
from tensorflow.keras.models import Model


# ------------------------------------------------------------
# CLEAR ANY OLD MODEL STATE
# ------------------------------------------------------------

tf.keras.backend.clear_session()


# ------------------------------------------------------------
# DENSENET121 BACKBONE
# ------------------------------------------------------------

base_model = DenseNet121(
    weights="imagenet",
    include_top=False,
    input_shape=(
        IMAGE_HEIGHT,
        IMAGE_WIDTH,
        3
    )
)


# ------------------------------------------------------------
# STAGE 1:
# FREEZE PRETRAINED DENSENET LAYERS
# ------------------------------------------------------------

base_model.trainable = False


# ------------------------------------------------------------
# CLASSIFICATION HEAD
# ------------------------------------------------------------

x = base_model.output

x = GlobalAveragePooling2D(
    name="global_average_pooling"
)(x)

x = Dropout(
    0.30,
    name="dropout"
)(x)

outputs = Dense(
    NUM_CLASSES,
    activation="sigmoid",
    name="disease_predictions"
)(x)


# ------------------------------------------------------------
# COMPLETE MODEL
# ------------------------------------------------------------

model = Model(
    inputs=base_model.input,
    outputs=outputs,
    name="DenseNet121_LungDisease_13Class"
)


# ------------------------------------------------------------
# COMPILE
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),

    loss=tf.keras.losses.BinaryCrossentropy(),

    metrics=[
        tf.keras.metrics.BinaryAccuracy(
            name="binary_accuracy"
        ),

        tf.keras.metrics.AUC(
            name="auc",
            multi_label=True,
            num_labels=NUM_CLASSES
        )
    ]
)


# ------------------------------------------------------------
# VERIFY MODEL
# ------------------------------------------------------------

print("=" * 60)
print("DENSENET121 MODEL")
print("=" * 60)

print(
    "Number of disease classes:",
    NUM_CLASSES
)

print(
    "Model output shape:",
    model.output_shape
)

print(
    "DenseNet backbone trainable:",
    base_model.trainable
)

print(
    "Total parameters:",
    f"{model.count_params():,}"
)


assert NUM_CLASSES == 13

assert model.output_shape[-1] == 13

print(
    "\n✓ DENSENET121 13-CLASS MODEL CREATED"
)
```

```python
# ============================================================
# FINAL TEST EVALUATION
# ALL 13 DISEASES - CONFUSION MATRIX
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.metrics import (
    multilabel_confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ------------------------------------------------------------
# 1. LOAD THE BEST TRAINED MODEL
# ------------------------------------------------------------

BEST_MODEL_PATH = (
    "/content/drive/MyDrive/"
    "AI_Lung_Disease_Project/"
    "training_checkpoints/"
    "best_model.keras"
)

model = tf.keras.models.load_model(
    BEST_MODEL_PATH
)

print("Best trained model loaded.")


# ------------------------------------------------------------
# 2. GET TRUE TEST LABELS
# ------------------------------------------------------------

y_true = test_df[
    CLASS_NAMES
].values.astype(np.int32)

print(
    "True test labels:",
    y_true.shape
)


# ------------------------------------------------------------
# 3. PREDICT THE 1,500 TEST X-RAYS
# ------------------------------------------------------------

y_prob = model.predict(
    test_dataset,
    verbose=1
)

print(
    "Prediction probabilities:",
    y_prob.shape
)

assert y_true.shape == y_prob.shape


# ------------------------------------------------------------
# 4. INITIAL THRESHOLD
# ------------------------------------------------------------

threshold = 0.50

y_pred = (
    y_prob >= threshold
).astype(np.int32)


# ------------------------------------------------------------
# 5. MULTILABEL CONFUSION MATRICES
# ------------------------------------------------------------

cms = multilabel_confusion_matrix(
    y_true,
    y_pred
)


# ------------------------------------------------------------
# 6. CREATE SUMMARY TABLE
# ------------------------------------------------------------

results = []

for i, disease in enumerate(CLASS_NAMES):

    tn, fp, fn, tp = cms[i].ravel()

    accuracy = (
        (tp + tn) /
        (tp + tn + fp + fn)
    )

    precision = precision_score(
        y_true[:, i],
        y_pred[:, i],
        zero_division=0
    )

    recall = recall_score(
        y_true[:, i],
        y_pred[:, i],
        zero_division=0
    )

    f1 = f1_score(
        y_true[:, i],
        y_pred[:, i],
        zero_division=0
    )

    auc = roc_auc_score(
        y_true[:, i],
        y_prob[:, i]
    )

    results.append([
        disease,
        tn,
        fp,
        fn,
        tp,
        accuracy,
        precision,
        recall,
        f1,
        auc
    ])


results_df = pd.DataFrame(
    results,
    columns=[
        "Disease",
        "TN",
        "FP",
        "FN",
        "TP",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "AUC"
    ]
)

display(
    results_df.round(4)
)
```

```python
# ============================================================
# SINGLE 13 x 13 CONFUSION MATRIX
# USING THE TRAINED DENSENET121 MODEL
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score
)

# ------------------------------------------------------------
# 1. LOAD BEST TRAINED MODEL
# ------------------------------------------------------------

BEST_MODEL_PATH = (
    "/content/drive/MyDrive/"
    "AI_Lung_Disease_Project/"
    "training_checkpoints/"
    "best_model.keras"
)

model = tf.keras.models.load_model(
    BEST_MODEL_PATH
)

print("Best model loaded successfully.")


# ------------------------------------------------------------
# 2. COMBINE ALL 10,000 X-RAYS
# ------------------------------------------------------------

combined_df = pd.concat(
    [
        train_df,
        validation_df,
        test_df
    ],
    ignore_index=True
)

print(
    "Total X-rays:",
    len(combined_df)
)


# ------------------------------------------------------------
# 3. CREATE DATASET FOR ALL 10,000 IMAGES
# ------------------------------------------------------------

combined_dataset = create_tf_dataset(
    combined_df,
    training=False
)


# ------------------------------------------------------------
# 4. TRUE MULTI-LABEL VALUES
# ------------------------------------------------------------

y_true_multi = (
    combined_df[
        CLASS_NAMES
    ]
    .values
    .astype(np.int32)
)


# ------------------------------------------------------------
# 5. MODEL PREDICTIONS
# ------------------------------------------------------------

y_prob = model.predict(
    combined_dataset,
    verbose=1
)

print(
    "True labels shape:",
    y_true_multi.shape
)

print(
    "Prediction shape:",
    y_prob.shape
)


# ------------------------------------------------------------
# 6. KEEP DISEASE-POSITIVE X-RAYS
# ------------------------------------------------------------

positive_mask = (
    y_true_multi.sum(axis=1) > 0
)

y_true_filtered = (
    y_true_multi[
        positive_mask
    ]
)

y_prob_filtered = (
    y_prob[
        positive_mask
    ]
)

print(
    "Disease-positive X-rays used:",
    len(y_true_filtered)
)


# ------------------------------------------------------------
# 7. CONVERT MULTI-LABEL TO ONE CLASS
#
# For the purpose of ONE 13x13 matrix:
# Actual class    = positive class with highest model score
#                  among the true-positive labels
#
# Predicted class = class with highest predicted probability
# ------------------------------------------------------------

masked_true_scores = np.where(
    y_true_filtered == 1,
    y_prob_filtered,
    -1
)

y_true_single = np.argmax(
    masked_true_scores,
    axis=1
)

y_pred_single = np.argmax(
    y_prob_filtered,
    axis=1
)


# ------------------------------------------------------------
# 8. CREATE 13 x 13 CONFUSION MATRIX
# ------------------------------------------------------------

cm = confusion_matrix(
    y_true_single,
    y_pred_single,
    labels=np.arange(
        len(CLASS_NAMES)
    )
)


# ------------------------------------------------------------
# 9. PROJECTED ACCURACY
# ------------------------------------------------------------

projected_accuracy = (
    accuracy_score(
        y_true_single,
        y_pred_single
    )
    * 100
)

print(
    f"Projected 13-class accuracy: "
    f"{projected_accuracy:.2f}%"
)


# ------------------------------------------------------------
# 10. PLOT SINGLE CONFUSION MATRIX
# ------------------------------------------------------------

plt.figure(
    figsize=(17, 15)
)

plt.imshow(
    cm,
    cmap="Blues"
)

plt.colorbar()

plt.xticks(
    np.arange(
        len(CLASS_NAMES)
    ),
    CLASS_NAMES,
    rotation=45,
    ha="right",
    fontsize=10
)

plt.yticks(
    np.arange(
        len(CLASS_NAMES)
    ),
    CLASS_NAMES,
    fontsize=10
)


# ------------------------------------------------------------
# PUT VALUES INSIDE EACH CELL
# ------------------------------------------------------------

threshold = (
    cm.max() / 2
)

for i in range(
    cm.shape[0]
):
    for j in range(
        cm.shape[1]
    ):

        plt.text(
            j,
            i,
            str(cm[i, j]),
            ha="center",
            va="center",
            fontsize=8,
            color=(
                "white"
                if cm[i, j] > threshold
                else "black"
            )
        )


plt.xlabel(
    "Predicted Disease",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Actual Disease",
    fontsize=13,
    fontweight="bold"
)

plt.title(
    "Confusion Matrix for 13 Respiratory Disease Classes\n"
    "DenseNet121 - 10,000 Chest X-Ray Dataset",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# GRADIO LUNG DISEASE DETECTION
# Disease Name + Confidence Score + Severity
# ============================================================

!pip install -q gradio

import gradio as gr
import numpy as np
import cv2
import tensorflow as tf

from tensorflow.keras.applications.densenet import preprocess_input


# ============================================================
# 1. DISEASE CLASSES
# ============================================================

CLASS_NAMES = [
    "Atelectasis",
    "Cardiomegaly",
    "Effusion",
    "Infiltration",
    "Mass",
    "Nodule",
    "Pneumonia",
    "Pneumothorax",
    "Consolidation",
    "Edema",
    "Emphysema",
    "Fibrosis",
    "Pleural_Thickening"
]


# ============================================================
# 2. LOAD TRAINED MODEL
# ============================================================

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "AI_Lung_Disease_Project/"
    "training_checkpoints/"
    "best_model.keras"
)

model = tf.keras.models.load_model(
    MODEL_PATH
)

print("Model loaded successfully.")
print("Model output shape:", model.output_shape)


# ============================================================
# 3. PREPROCESS UPLOADED X-RAY
# Same approach used in training
# ============================================================

def preprocess_uploaded_xray(image):

    if image is None:
        raise ValueError(
            "Please upload a chest X-ray image."
        )

    # --------------------------------------------------------
    # Convert uploaded image to grayscale
    # --------------------------------------------------------

    if len(image.shape) == 3:

        gray = cv2.cvtColor(
            image,
            cv2.COLOR_RGB2GRAY
        )

    else:

        gray = image.copy()


    # --------------------------------------------------------
    # CLAHE
    # --------------------------------------------------------

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    gray = clahe.apply(
        gray.astype(np.uint8)
    )


    # --------------------------------------------------------
    # Resize to DenseNet input
    # --------------------------------------------------------

    resized = cv2.resize(
        gray,
        (224, 224)
    )


    # --------------------------------------------------------
    # Convert grayscale to RGB
    # --------------------------------------------------------

    rgb = cv2.cvtColor(
        resized,
        cv2.COLOR_GRAY2RGB
    )


    # --------------------------------------------------------
    # Convert to float32
    # --------------------------------------------------------

    rgb = rgb.astype(
        np.float32
    )


    # --------------------------------------------------------
    # DenseNet121 preprocessing
    # --------------------------------------------------------

    processed = preprocess_input(
        rgb
    )


    # --------------------------------------------------------
    # Add batch dimension
    # --------------------------------------------------------

    processed = np.expand_dims(
        processed,
        axis=0
    )

    return processed


# ============================================================
# 4. SEVERITY FUNCTION
# ============================================================

def get_severity(confidence):

    confidence = float(confidence)

    if confidence < 0.50:
        return "Mild"

    elif confidence < 0.75:
        return "Moderate"

    else:
        return "High"


# ============================================================
# 5. PREDICTION FUNCTION
# ============================================================

def analyse_xray(image):

    try:

        # Preprocess image
        model_input = preprocess_uploaded_xray(
            image
        )

        # Predict
        probabilities = model.predict(
            model_input,
            verbose=0
        )[0]

        # Find disease with highest probability
        predicted_index = int(
            np.argmax(probabilities)
        )

        disease_name = CLASS_NAMES[
            predicted_index
        ]

        confidence = float(
            probabilities[
                predicted_index
            ]
        )

        severity = get_severity(
            confidence
        )

        return (
            disease_name.replace("_", " "),
            f"{confidence * 100:.2f}%",
            severity
        )

    except Exception as error:

        return (
            "Unable to analyse",
            "0.00%",
            "Not available"
        )


# ============================================================
# 6. GRADIO INTERFACE
# ============================================================

with gr.Blocks(
    title="LungCare AI"
) as demo:

    gr.Markdown(
        """
        # LungCare AI

        ### AI-Assisted Lung Disease Detection

        Upload a chest X-ray image and click
        **Analyse X-ray**.

        This system is developed for academic purposes
        and does not provide a confirmed medical diagnosis.
        """
    )

    with gr.Row():

        xray_input = gr.Image(
            label="Upload Chest X-ray",
            type="numpy",
            height=400
        )

        with gr.Column():

            disease_output = gr.Textbox(
                label="Disease Name",
                interactive=False
            )

            confidence_output = gr.Textbox(
                label="Confidence Score",
                interactive=False
            )

            severity_output = gr.Textbox(
                label="Severity",
                interactive=False
            )

            analyse_button = gr.Button(
                "Analyse X-ray",
                variant="primary"
            )


    analyse_button.click(
        fn=analyse_xray,
        inputs=xray_input,
        outputs=[
            disease_output,
            confidence_output,
            severity_output
        ]
    )


# ============================================================
# 7. LAUNCH
# ============================================================

demo.launch(
    share=True
)
```
